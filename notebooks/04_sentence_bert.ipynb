# 04 Model B: Sentence-BERT
`all-MiniLM-L6-v2`, pretrained, not fine-tuned. Needs `pip install -r requirements.txt` and internet the first time.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))
import numpy as np, pandas as pd
pd.set_option("display.max_colwidth", 110)
from src.sentence_bert_model import SbertUnavailable, SentenceBertModel
m = SentenceBertModel()
try:
    v = m.encode(['Garbage has not been collected from my street for three days',
                  'The waste collection vehicle has not visited our locality since Monday',
                  'The streetlight near the park is not working'])
    print('embedding shape:', v.shape)
    print('paraphrase similarity:', round(float(v[0] @ v[1]), 3), ' unrelated:', round(float(v[0] @ v[2]), 3))
except SbertUnavailable as e:
    print('Sentence-BERT is not available on this machine:', e)
```
